> *Tips: jika muncul `RuntimeWarning: overflow`, itu adalah bagian dari hasil eksperimen. Jelaskan artinya.*

````python
# Jawaban A5
# Implementasi Class LinearRegressionGD
class LinearRegressionGD:

  def __init__(self, lr=0.01, n_iterations=5000):
    self.lr = lr
    self.n_iterations = n_iterations
    self.losses = []

  def fit(self, X, y):
    n_samples, n_features = X.shape
    self.w = np.zeros(n_features)
    self.b = 0
    self.losses = []

    for _ in range(self.n_iterations):
      y_pred = np.dot(X, self.w) + self.b
      loss = np.mean((y - y_pred) ** 2)
      self.losses.append(loss)

      # Perhitungan gradien
      dw = (-2 / n_samples) * np.dot(X.T, (y - y_pred))
      db = (-2 / n_samples) * np.sum(y - y_pred)

      # Pembaruan parameter
      self.w -= self.lr * dw
      self.b -= self.lr * db


# 3 nilai learning rate yang diuji
learning_rates = [0.001, 0.02, 0.1]
models_gd = {}

plt.figure(figsize=(10, 6))

for lr in learning_rates:
  model_gd = LinearRegressionGD(lr=lr, n_iterations=5000)
  model_gd.fit(X_train_a, y_train_a)
  models_gd[lr] = model_gd

  print(
      f'LR = {lr:<5} | Intercept (b): {model_gd.b:.4f} | Slope (m):'
      f' {model_gd.w[0]:.4f}'
  )

  # Plotting grafik loss (MSE) vs iterasi
  plt.plot(
      range(len(model_gd.losses)), model_gd.losses, label=f'Learning Rate = {lr}'
  )

plt.title('Grafik Loss (MSE) vs Iterasi untuk Berbagai Learning Rate')
plt.xlabel('Iterasi')
plt.ylabel('Loss (MSE) - Skala Log')
plt.yscale('log')
plt.legend()
plt.grid(True)
plt.show()
````

**Jawaban A5 (analisis):**lr = 0.001 (Terlalu Lambat): Langkah pembaruan parameter terlalu kecil. Meskipun kurva loss terus menurun secara stabil, penurunannya sangat lambat sehingga belum mencapai titik konvergensi optimal dalam batas 5000 iterasi.

lr = 0.02 (Pas / Tepat): Ukuran langkah pembaruan bobot sangat ideal. Kurva loss menurun secara konsisten dengan cepat dan stabil mencapai titik minimum, menghasilkan nilai intercept dan koefisien yang hampir sama persis dengan hasil Normal Equation pada Soal A3.

### Soal B2 – Split Data & Training
1. Bagi data dengan `test_size=0.25` dan `random_state=RANDOM_STATE`.
2. Latih model `LogisticRegression` (gradient descent, dari materi) dengan `learning_rate=0.1` dan `n_iterations=10000`.
3. Tampilkan intercept ($b$) dan koefisien ($m_1, m_2$), lalu tuliskan persamaannya:
   $$P(\text{Spam}) = \sigma(b + m_1 \cdot \text{JumlahLink} + m_2 \cdot \text{KataPromo})$$
4. Buat grafik **log loss vs iterasi**. Apakah loss sudah turun dan stabil?

````python
# Jawaban B2
# Implementasi Class LogisticRegression dengan Gradient Descent
class LogisticRegression:

  def __init__(self, lr=0.1, n_iterations=10000):
    self.lr = lr
    self.n_iterations = n_iterations
    self.losses = []

  def sigmoid(self, z):
    return 1 / (1 + np.exp(-np.clip(z, -250, 250)))

  def fit(self, X, y):
    n_samples, n_features = X.shape
    self.w = np.zeros(n_features)
    self.b = 0
    self.losses = []

    for _ in range(self.n_iterations):
      linear_model = np.dot(X, self.w) + self.b
      y_predicted = self.sigmoid(linear_model)

      # Binary Cross Entropy Loss (Log Loss)
      epsilon = 1e-15
      y_pred_clipped = np.clip(y_predicted, epsilon, 1 - epsilon)
      loss = -np.mean(
          y * np.log(y_pred_clipped) + (1 - y) * np.log(1 - y_pred_clipped)
      )
      self.losses.append(loss)

      # Gradien
      dw = (1 / n_samples) * np.dot(X.T, (y_predicted - y))
      db = (1 / n_samples) * np.sum(y_predicted - y)

      # Pembaruan parameter
      self.w -= self.lr * dw
      self.b -= self.lr * db

  def predict_proba(self, X):
    linear_model = np.dot(X, self.w) + self.b
    return self.sigmoid(linear_model)

  def predict(self, X, threshold=0.5):
    return (self.predict_proba(X) >= threshold).astype(int)


# 1. Bagi data dengan test_size=0.25 dan random_state=RANDOM_STATE
X_train_b, X_test_b, y_train_b, y_test_b = train_test_split(
    X_email, y_email, test_size=0.25, random_state=RANDOM_STATE
)

# 2. Latih model LogisticRegression dengan learning_rate=0.1 dan n_iterations=10000
model_log = LogisticRegression(lr=0.1, n_iterations=10000)
model_log.fit(X_train_b, y_train_b)

# 3. Tampilkan intercept (b) dan koefisien (m1, m2)
print(f'Intercept (b)           : {model_log.b:.4f}')
print(f'Koefisien m1 (JumlahLink): {model_log.w[0]:.4f}')
print(f'Koefisien m2 (KataPromo) : {model_log.w[1]:.4f}')
print(
    f'Persamaan: P(Spam) = sigma({model_log.b:.4f} +'
    f' {model_log.w[0]:.4f}*JumlahLink + {model_log.w[1]:.4f}*KataPromo)'
)

# 4. Buat grafik log loss vs iterasi
plt.figure(figsize=(8, 4))
plt.plot(range(len(model_log.losses)), model_log.losses, color='green')
plt.title('Grafik Log Loss vs Iterasi (Logistic Regression)')
plt.xlabel('Iterasi')
plt.ylabel('Log Loss')
plt.grid(True)
plt.show()
````

**Jawaban B2 (analisis):**Apakah loss sudah turun dan stabil? Ya, berdasarkan grafik Log Loss vs Iterasi, nilai loss mengalami penurunan yang sangat tajam pada iterasi awal dan secara perlahan menjadi datar (konvergen) mendekati nilai minimum seiring bertambahnya iterasi.

Kesimpulan: Hal ini menandakan bahwa proses gradient descent berjalan dengan sangat baik, stabil, dan parameter model telah berhasil dioptimalkan.

### Soal B3 – Evaluasi Model
Selain `accuracy_score` dan `confusion_matrix` dari materi, **implementasikan sendiri** fungsi berikut:

$$
Precision = \frac{TP}{TP + FP} \qquad
Recall = \frac{TP}{TP + FN} \qquad
F1 = 2 \cdot \frac{Precision \cdot Recall}{Precision + Recall}
$$

1. Hitung accuracy, confusion matrix, precision, recall, dan F1-score pada **data testing**.
2. Jelaskan arti nilai **FP** (*false positive*) dan **FN** (*false negative*) pada kasus filter spam. Kesalahan mana yang menurut Anda **lebih merugikan pengguna email**? Mengapa?

````python
# Jawaban B3
# 1. Implementasi fungsi evaluasi klasifikasi sendiri menggunakan NumPy
def confusion_matrix_custom(y_true, y_pred):
  tp = np.sum((y_true == 1) & (y_pred == 1))
  tn = np.sum((y_true == 0) & (y_pred == 0))
  fp = np.sum((y_true == 0) & (y_pred == 1))
  fn = np.sum((y_true == 1) & (y_pred == 0))
  return np.array([[tn, fp], [fn, tp]])


def accuracy_custom(y_true, y_pred):
  return np.mean(y_true == y_pred)


def precision_custom(y_true, y_pred):
  tp = np.sum((y_true == 1) & (y_pred == 1))
  fp = np.sum((y_true == 0) & (y_pred == 1))
  return tp / (tp + fp) if (tp + fp) > 0 else 0.0


def recall_custom(y_true, y_pred):
  tp = np.sum((y_true == 1) & (y_pred == 1))
  fn = np.sum((y_true == 1) & (y_pred == 0))
  return tp / (tp + fn) if (tp + fn) > 0 else 0.0


def f1_custom(y_true, y_pred):
  prec = precision_custom(y_true, y_pred)
  rec = recall_custom(y_true, y_pred)
  return 2 * (prec * rec) / (prec + rec) if (prec + rec) > 0 else 0.0


# 2. Menghitung metrik evaluasi pada data testing
y_pred_b = model_log.predict(X_test_b)

cm = confusion_matrix_custom(y_test_b, y_pred_b)
acc = accuracy_custom(y_test_b, y_pred_b)
prec = precision_custom(y_test_b, y_pred_b)
rec = recall_custom(y_test_b, y_pred_b)
f1 = f1_custom(y_test_b, y_pred_b)

print('Confusion Matrix (TN, FP / FN, TP):')
print(cm)
print(f'Accuracy  : {acc:.4f}')
print(f'Precision : {prec:.4f}')
print(f'Recall    : {rec:.4f}')
print(f'F1-Score  : {f1:.4f}')
````

**Jawaban B3 (analisis):**Arti Nilai FP (False Positive): Kondisi di mana email penting atau normal (Bukan Spam) secara keliru diprediksi oleh sistem sebagai Spam.

Arti Nilai FN (False Negative): Kondisi di mana email berjenis Spam gagal terdeteksi oleh sistem dan lolos masuk ke kotak masuk (Inbox) pengguna.

Kesalahan mana yang lebih merugikan pengguna email? Kesalahan False Positive (FP) jauh lebih merugikan.

Mengapa? Jika email penting (seperti surat panggilan kerja, informasi akademik, atau transaksi pembayaran) masuk ke folder spam dan terlewat oleh pengguna, hal itu dapat menimbulkan dampak negatif yang besar. Sebaliknya, jika terjadi False Negative, pengguna hanya perlu menghapus email spam tersebut secara manual dari inbox tanpa kehilangan informasi penting.

### Soal B4 – Decision Boundary
1. Hitung persamaan **decision boundary** (garis saat probabilitas = 0.5):
   $$b + m_1 x_1 + m_2 x_2 = 0 \;\Rightarrow\; x_2 = -\frac{b + m_1 x_1}{m_2}$$
2. Buat visualisasi data **testing** beserta *background* probabilitas (`contourf`) dan garis decision boundary, seperti di materi.
3. Tunjukkan (beri tanda/penjelasan) data testing mana saja yang **salah diklasifikasikan**.

````python
# Jawaban B4
# 1. Menghitung dan menampilkan persamaan Decision Boundary (P = 0.5)
# Formula: b + m1*x1 + m2*x2 = 0  =>  x2 = -(b + m1*x1) / m2
b_val = model_log.b
m1_val = model_log.w[0]
m2_val = model_log.w[1]

print('=== PERSAMAAN DECISION BOUNDARY (P = 0.5) ===')
print(
    f'KataPromo (x2) = -({b_val:.4f} + {m1_val:.4f} * JumlahLink) /'
    f' {m2_val:.4f}'
)

# 2. Visualisasi Decision Boundary, Background Probabilitas (Contour), dan Data Testing
plt.figure(figsize=(10, 6))

# Membuat grid meshgrid untuk kontur warna probabilitas
x1_min, x1_max = X_email[:, 0].min() - 1, X_email[:, 0].max() + 1
x2_min, x2_max = X_email[:, 1].min() - 1, X_email[:, 1].max() + 1
xx1, xx2 = np.meshgrid(
    np.linspace(x1_min, x1_max, 200), np.linspace(x2_min, x2_max, 200)
)

# Menghitung probabilitas untuk seluruh area grid
grid_points = np.c_[xx1.ravel(), xx2.ravel()]
Z = model_log.predict_proba(grid_points).reshape(xx1.shape)

# Plot kontur background probabilitas
plt.contourf(xx1, xx2, Z, alpha=0.3, cmap='coolwarm')
plt.colorbar(label='Probabilitas Spam')

# Plot Garis Decision Boundary (P = 0.5)
x1_line = np.linspace(x1_min, x1_max, 100)
x2_line = -(b_val + m1_val * x1_line) / m2_val
plt.plot(
    x1_line, x2_line, color='black', linestyle='--', linewidth=2, label='Decision Boundary (P = 0.5)'
)

# Plot Data Testing Aktual (Spam vs Bukan Spam)
plt.scatter(
    X_test_b[y_test_b == 1, 0],
    X_test_b[y_test_b == 1, 1],
    color='red',
    label='Spam (1)',
    marker='o',
    s=60
)
plt.scatter(
    X_test_b[y_test_b == 0, 0],
    X_test_b[y_test_b == 0, 1],
    color='blue',
    label='Bukan Spam (0)',
    marker='s',
    s=60
)

# 3. Menandai data testing yang salah diklasifikasikan (Misclassified)
y_pred_b = model_log.predict(X_test_b)
misclassified = y_test_b != y_pred_b

if np.sum(misclassified) > 0:
  plt.scatter(
      X_test_b[misclassified, 0],
      X_test_b[misclassified, 1],
      color='yellow',
      s=180,
      marker='X',
      edgecolor='black',
      linewidth=1.5,
      label='Salah Klasifikasi (Misclassified)',
  )
  print(f'\nJumlah data testing yang salah diklasifikasikan: {np.sum(misclassified)} data')
else:
  print('\nSemua data testing terklasifikasi dengan benar (Akurasi 100%).')

plt.title('Decision Boundary Logistic Regression & Kontur Probabilitas (Data Testing)')
plt.xlabel('Jumlah Link (x1)')
plt.ylabel('Jumlah Kata Promo (x2)')
plt.legend(loc='upper right')
plt.grid(True)
plt.show()
````

Dalam variabel fitur, persamaannya menjadi:$$\text{JumlahKataPromo } (x_2) = -\frac{b + m_1 \cdot \text{JumlahLink}}{m_2}$$Penjelasan Data yang Salah Diklasifikasikan: Titik data yang ditandai dengan ikon tanda silang 'X' berwarna kuning merupakan titik data testing yang salah diklasifikasikan oleh model.
